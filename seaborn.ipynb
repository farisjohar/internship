```python
df = sns.load_dataset('tips')
```

```python
df
```

```python
sns.scatterplot(data=df, x='total_bill', y='tip', hue='day')
plt.title('Scatter Polt:Total Bill vs Tip')
plt.show()
```

```python
sns.lineplot(data=df, x='total_bill', y='tip')

plt.title("Line Plot: Total Bill vs Tip")
plt.show()
```

```python
sns.barplot(data=df, x='day', y='total_bill')
plt.title('Average Total Bill by Data')
plt.show()
```

```python
sns.boxplot(data=df,x='day',y='total_bill',hue='day')
plt.title('average total bill by day')
plt.show()
```

```python
sns.kdeplot(data=df, x='total_bill', fill=True, color='blue')
plt.title("Kernel Density Estimate (KDE) of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Density")
plt.show()
```

```python
sns.violinplot(data=df, x='day',y='total_bill')
plt.title("violin plot:total Bill distribution by day")
plt.show()
```

```python
df = sns.load_dataset('tips')
sns.histplot(data=df, x='total_bill', kde=True)
plt.title("Distribution Plot: Total Bill")
plt.show()
```

```python
sns.pairplot(df, hue="sex")
plt.show()
```

```python
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```
